**Assignment 2: Accessing Tuple Elements**

**Print the first, middle, and last elements of the tuple created in Assignment 1.**

```python
first_element = tuple[0]
print("first element: ", first_element)
middle_element = tuple[len(tuple)//2]
print("Middle Element: ", middle_element)
last_element = tuple[-1]
print("Last Element: ",last_element)
```

**Assignment 4: Nested Tuples**

**Create a nested tuple representing a 3x3 matrix and print the matrix. Access and print the element at the second row and third column.**

```python
matrix = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 9)
)
print("3x3 matrix \n")
for row in matrix:
    print(row)
element = matrix[1][2]
print("\nElement at second row, third column:",element)
```

**Assignment 6: Tuple Methods**

**Create a tuple with duplicate elements and count the occurrences of an element. Find the index of the first occurrence of an element in the tuple.**

```python
my_tuple = (10,10, 20, 30, 20, 40, 20, 50)
count_20 = my_tuple.count(20)
index_20 = my_tuple.index(20)
print("Original Tuple:", my_tuple)
print("Count of 20:", count_20)
print("Index of first occurrence of 20:", index_20)
```

**Assignment 10: Tuple and List**
